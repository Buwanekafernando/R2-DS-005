```python
import pandas as pd
import os
```

```python
RAW_DIR = "data/raw/"
```

```python
# List of your 7 category files with their intended labels
CATEGORY_FILES = {
    "amazon_reviews_us_Beauty_v1_00.tsv":       {"label": 1, "label_name": "System1"},
    "amazon_reviews_us_Apparel_v1_00.tsv":       {"label": 1, "label_name": "System1"},
    "amazon_reviews_us_Grocery_v1_00.tsv":       {"label": 1, "label_name": "System1"},
    "amazon_reviews_us_Pet_Products_v1_00.tsv":  {"label": 1, "label_name": "System1"},
    "amazon_reviews_us_Electronics_v1_00.tsv":   {"label": 0, "label_name": "System2"},
    "amazon_reviews_us_Baby_v1_00.tsv":          {"label": None, "label_name": "Mixed"},
    "amazon_reviews_us_Sports_v1_00.tsv":        {"label": None, "label_name": "Mixed"},
}
```

```python
# Explore each file - check size, columns, missing values
for filename, meta in CATEGORY_FILES.items():
    filepath = os.path.join(RAW_DIR, filename)
    if not os.path.exists(filepath):
        print(f"MISSING: {filename}")
        continue

    # Read only first 1000 rows for exploration (files are huge)
    df = pd.read_csv(filepath, sep='\t', nrows=1000,
                     on_bad_lines='skip', low_memory=False)

    print(f"\n{'='*55}")
    print(f"FILE: {filename}")
    print(f"Intended Label: {meta['label_name']}")
    print(f"Columns: {df.columns.tolist()}")
    print(f"Shape (sample): {df.shape}")
    print(f"Missing review_body: {df['review_body'].isna().sum()}")
    print(f"Missing product_title: {df['product_title'].isna().sum()}")
    print(f"Star rating distribution:\n{df['star_rating'].value_counts().sort_index()}")
    print(f"\nSample review_body:\n{df['review_body'].iloc[0][:200]}")
```
